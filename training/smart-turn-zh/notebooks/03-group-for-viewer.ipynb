# Group Files for Viewer

For each WAV in `data/wav/` that has a matching VAD `.npy` and ASR `.json`,
move all three into `data/grouped/{stem}/` so they can be opened together in the viewer.

```python
import shutil
from pathlib import Path

WAV_DIR = Path("../data/wav")
VAD_DIR = Path("../data/vad_probs")
ASR_DIR = Path("../data/asr_results")
OUT_DIR = Path("../data/grouped")

OUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# Preview — show what will be moved vs skipped
to_group = []
skipped = []

for wav_path in sorted(WAV_DIR.glob("*.wav")):
    stem = wav_path.stem
    vad_path = VAD_DIR / f"{stem}.npy"
    asr_path = ASR_DIR / f"{stem}.json"

    if not vad_path.exists() or not asr_path.exists():
        missing = []
        if not vad_path.exists(): missing.append("VAD")
        if not asr_path.exists(): missing.append("ASR")
        skipped.append((stem, missing))
    else:
        to_group.append((wav_path, vad_path, asr_path))
        print(f"  ✓ {stem}")

print(f"\nReady to group: {len(to_group)}")
if skipped:
    print(f"Skipped: {len(skipped)}")
    for stem, missing in skipped:
        print(f"  ✗ {stem} — missing {', '.join(missing)}")
```

```python
# Move files
for wav_path, vad_path, asr_path in to_group:
    dest = OUT_DIR / wav_path.stem
    dest.mkdir(exist_ok=True)

    shutil.move(str(wav_path), dest / wav_path.name)
    shutil.move(str(vad_path), dest / vad_path.name)
    shutil.move(str(asr_path), dest / asr_path.name)

    print(f"  {wav_path.stem}/")

print(f"\nMoved: {len(to_group)}")
```

```python
# Verify
for d in sorted(OUT_DIR.iterdir()):
    if d.is_dir():
        files = sorted(f.name for f in d.iterdir())
        print(f"{d.name}/  {files}")
```
